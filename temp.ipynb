```python
from pathlib import Path
import tarfile, datetime, zoneinfo
import gzip, zipfile, json
```

```python
RAW_PATH = Path("datastore/raw")
```

```python
pages = []
revisions = []
events = []
labels = []
manifest = []

with zipfile.ZipFile(RAW_PATH / 'full-wiki-logs.zip', "r") as zip:
    print(zip.namelist())
    with zip.open('pages.jsonl') as f:
        for line in f:
            data = json.loads(line)
            pages.append(data)

    with zip.open('revisions.jsonl') as f:
        for line in f:
            data = json.loads(line)
            revisions.append(data)

    with zip.open('events.jsonl') as f:
        for line in f:
            data = json.loads(line)
            events.append(data)

    with zip.open('labels.jsonl') as f:
        for line in f:
            data = json.loads(line)
            labels.append(data)

    with zip.open('manifest.json') as f:
        manifest = json.load(f)
```
